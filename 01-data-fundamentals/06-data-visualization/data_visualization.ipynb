## Explore BMI

### 9\) Regplot - Charges vs. BMI

9a) Make a seaborn regplot to visualize the relationship between charges and bmi for the entire dataset

```python
# Make a seaborn regplot for bmi vs. charges.
fig , ax = plt.subplots()
kws_line = dict(color = "black")
scatter_kws = dict(edgecolor="white")
ax = sns.regplot(data=df_insurance , x="bmi" , y="charges" ,line_kws=kws_line ,scatter_kws=scatter_kws)
ax.set(xlabel="BMI" , ylabel="Charges" , title="Relationship Between BMU and Insurance Charges") ;

```

### 11\)  Linear Model Plot - Charges vs. BMI - Colored by Smoker



11a) Compare the relationship of bmi and charges for smokers and nonsmokers using seaborn's `lmplot` to calculate a separate trendline for Smokers vs. Non-Smokers.


```python
# Create an lmplot of bmi vs. charges, colored by smoker.
scatter_kws = dict(edgecolor="white")
kws_line = dict(color="black")
g = sns.lmplot(data=df_insurance , x="bmi" , y="charges" , hue="smoker" , scatter_kws=scatter_kws , line_kws=kws_line , height=6, aspect=1.66)
g.set(xlabel="BMI",ylabel="Charges" , title="Regression Analysis of BMI and Charges by Smoker Status") ;
```

11b) What do you notice from the visualization? (Compare the trendlines.)

- Your Answer:
  - `lmplot chart provides a clear statistical confirmation that impact of BMI on insurance charges is heavily dependent on smoking status. For non-smokers, regression line is nearly horizontal, suggesting that BMI is not the primary driver of high charges. In contrast, steep positive slope for smokers illustrates smoking not only raises the baseline insurance cost but also makes those charges highly sensitive to BMI increases. combination of smoking and high BMI leads to the most extreme financial outcomes in the dataset.`

### 14\) Bar Chart of Ratios (%) of Smokers in Each Region

14a) Calculate the ratio of smokers in each region and save it as a variable called "percent_smoker_by_region". (Hint: use .groupby with "region" combined with slicing the "smoker" column and end with .mean() )

```python
# use groupby to get the mean of smoker grouped by region
percent_smoker_by_region = df_insurance.groupby("region")["smoker"].mean().sort_values(ascending=False) * 100
percent_smoker_by_region
```

14b) Plot the ratios of smokers by region as a bar plot.



```python
# Plot the ratios (the results of then groupby above)
fig , ax = plt.subplots()
ax = sns.barplot(x=percent_smoker_by_region.index , y=percent_smoker_by_region.values)
ax.yaxis.set_major_formatter(mtick.PercentFormatter())
ax.set(xlabel="Region" , ylabel="Smoker Percentage" , title="Percentage of Smokers by Region") ;
```

  - `southeast`
