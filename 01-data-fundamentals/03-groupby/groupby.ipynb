```python
insurance_url = "/content/drive/MyDrive/AXSOSACADEMY/01-Fundamentals/Week02/Data/insurance_mod.csv"
```

```python
import pandas as pd

df_insurance = pd.read_csv(insurance_url)
```

```python
# 1 - Average age

avg_age = df_insurance["age"].mean()
print(f"Average age of this data set is {avg_age:.2f}")
```

```python
#2 Maximum age
max_age = df_insurance["age"].max()
max_age
```

```python
# 3 Min age
min_age = df_insurance["age"].min()
min_age
```

```python
# 4 average of charges

avg_charges = df_insurance['charges'].mean()
print(f" Average Charges :{avg_charges:.2f}")
```

```python
# 5 - average of charges grouped by sex

avg_charge_sex = df_insurance.groupby("sex")["charges"].mean().round(2)
avg_charge_sex
```

```python
# 6 - group by and smoker

group_sex_smoker = df_insurance.groupby("smoker")["charges"].mean().round(2)
group_sex_smoker
```

```python
# 7 - avg charges by smoker and sex

avg_cahrges_smoker_sex = df_insurance.groupby(["sex" , "smoker"])["charges"].mean().round(2)
avg_cahrges_smoker_sex
```
