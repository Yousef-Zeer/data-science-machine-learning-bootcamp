```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

```

```python
url = "/content/drive/MyDrive/AXSOSACADEMY/01-Fundamentals/Week03/Data/data-cleaning-core.csv"
df = pd.read_csv(url)

df.head()
```

## number of passengers who Survived vs. Died.

```python
number_passengers = df['Survived'].value_counts()
number_passengers
```

```python
fig , ax = plt.subplots()
number_passengers.index = ["Died" , "Survived"]
number_passengers.plot(kind="bar" , rot=360 , color=['#e74c3c', '#2ecc71'])
ax.set_xlabel("Passengers' Fate")
ax.set_ylabel("No of Passengers")
ax.set_title("Titanic Survival Distribution")
plt.show()
```

## Distribution of Age

```python
fig , ax = plt.subplots()
ax = sns.histplot(data=df , x="Age" , bins=20 , binwidth=5)
ax.set(xlabel="Age(Years)" , ylabel="Number of People" , title="Age Distribution") ;
```

```python
# The code for the last branch is working, but the analysis is unconvincing. I realized our use of -1 for missing ages is distorting the final data
ax = sns.histplot(data=df[df['Age'] > 0], x="Age", bins=20, color="skyblue" , binwidth=5)
```

## Fare Boxplot

```python
fig , ax = plt.subplots(figsize=(10,4))
data = df['Fare']
ax.boxplot(data , vert=False)
plt.xticks(np.arange(0, 551, 25))
ax.set( xlabel="Fare Amount ($)", ylabel="" , title="Distribution of Fare Prices") ;
```
