## 1. Environment Setup & Data Initialization
In this section, we import the necessary packages/Modules and define temperature datasets .



```python
# importing the required packages
import numpy as np
from statistics import mode

# Lists of Fahrenheit temps
week_1 = [45, 55, 49, 60, 52, 55, 58]
week_2 = [51, 49, 62, 50, 53, 56, 59]
week_3 = [60, 60, 61, 62, 57, 59, 58]
week_4 = [59, 63, 62, 61, 65, 66, 62]
```

## 2. Preliminary Results
Using functions from imported packages to find the measures for week_1 .

```python
print(f"Mean: {np.mean(week_1)}")
print(f"Median: {np.median(week_1)}")
print(f"Mode: {mode(week_1)}")
```

## 3. Custom Function Implementation
Creating a custom function `central_tendency` that returns the three measurements  

```python
# Custom function to calculate these measures

def central_tendency (temp):
    mean_temp = np.mean(temp)
    median_temp = np.median(temp)
    mode_temp = mode(temp)

    return mean_temp , median_temp , mode_temp
```

```python
central_tendency(week_1)
```

## 4. Enhanced Function
Modifying the function to include a keyword argument `verbose`

```python
#Modified function

def central_tendency (temp , verbose = True):
    mean_temp = np.mean(temp)
    median_temp = np.median(temp)
    mode_temp = mode(temp)

    if verbose :
        print(f"the mean is {mean_temp}\nthe median is {median_temp}\nthe mode is {mode_temp}")

    return mean_temp , median_temp , mode_temp
```

## 5. Test Cases
Testing the modified function by changing the arguments

```python
# Test Cases

central_tendency(week_2 , verbose=True)
central_tendency(week_2)
central_tendency(week_2 , verbose=False)
```
